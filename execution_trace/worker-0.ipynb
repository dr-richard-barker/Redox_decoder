```python
import pandas as pd
import numpy as np
import json

# Load metadata
meta = pd.read_csv('/mnt/shared-workspace/shared/preprocessing/sample_metadata_final.csv')
print(f'Total samples: {len(meta)}')
print(f'Columns: {list(meta.columns)}')

# --- Parse stimulus multi-labels ---
print('\n=== Stimulus Analysis ===')
all_stimuli = set()
for s in meta['stimulus'].dropna():
    parts = [p.strip() for p in str(s).split(';')]
    all_stimuli.update(parts)
all_stimuli = sorted(all_stimuli)
print(f'Unique individual stimuli: {len(all_stimuli)}')
for i, s in enumerate(all_stimuli):
    n = sum(1 for x in meta['stimulus'].dropna() if s in str(x).split(';'))
    print(f'  {i:2d}. {s}: {n} samples')

# --- Parse tissue multi-labels ---
print('\n=== Tissue Analysis ===')
all_tissues = set()
for t in meta['tissue'].dropna():
    parts = [p.strip() for p in str(t).split(';')]
    all_tissues.update(parts)
all_tissues = sorted(all_tissues)
print(f'Unique individual tissues: {len(all_tissues)}')
for i, t in enumerate(all_tissues):
    n = sum(1 for x in meta['tissue'].dropna() if t in str(x).split(';'))
    print(f'  {i:2d}. {t}: {n} samples')

# --- Build encoding maps ---
stimulus_to_idx = {s: i for i, s in enumerate(all_stimuli)}
tissue_to_idx = {t: i for i, t in enumerate(all_tissues)}

cond_dim = len(all_stimuli) + len(all_tissues)
print(f'\n=== Conditioning Dimension ===')
print(f'Stimulus dim: {len(all_stimuli)}')
print(f'Tissue dim: {len(all_tissues)}')
print(f'Total conditioning dim: {cond_dim}')

# --- Build multi-hot encoding matrix ---
print('\n=== Building multi-hot encoding ===')
cond_matrix = np.zeros((len(meta), cond_dim), dtype=np.float32)
for i, row in meta.iterrows():
    if pd.notna(row['stimulus']):
        for s in str(row['stimulus']).split(';'):
            s = s.strip()
            if s in stimulus_to_idx:
                cond_matrix[i, stimulus_to_idx[s]] = 1.0
    if pd.notna(row['tissue']):
        for t in str(row['tissue']).split(';'):
            t = t.strip()
            if t in tissue_to_idx:
                cond_matrix[i, len(all_stimuli) + tissue_to_idx[t]] = 1.0

print(f'Conditioning matrix: {cond_matrix.shape}')
print(f'Avg active dims per sample: {cond_matrix.sum(axis=1).mean():.2f}')
print(f'Samples with no stimulus label: {(cond_matrix[:, :len(all_stimuli)].sum(axis=1) == 0).sum()}')
print(f'Samples with no tissue label: {(cond_matrix[:, len(all_stimuli):].sum(axis=1) == 0).sum()}')

# Save encoding maps
encoding_info = {
    'stimulus_labels': all_stimuli,
    'tissue_labels': all_tissues,
    'stimulus_to_idx': stimulus_to_idx,
    'tissue_to_idx': tissue_to_idx,
    'cond_dim': cond_dim,
    'n_stimulus': len(all_stimuli),
    'n_tissue': len(all_tissues),
}
with open('/workspace/conditioning_encoding.json', 'w') as f:
    json.dump(encoding_info, f, indent=2)
np.save('/workspace/conditioning_matrix.npy', cond_matrix)
print('\nSaved: conditioning_encoding.json, conditioning_matrix.npy')
```

```python
import sys, os
sys.path.insert(0, '/workspace')
from cvae_model import CVAE, load_data, ExpressionDataset
import torch
import numpy as np
import pandas as pd
import json
import umap
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from matplotlib import rcParams

# Set font and SVG settings
rcParams['font.family'] = ['Liberation Sans', 'Arimo', 'DejaVu Sans']
rcParams['svg.fonttype'] = 'none'

os.makedirs('/workspace/cvae_model', exist_ok=True)
os.makedirs('/mnt/results/figures', exist_ok=True)

# Load data
print('=== Loading data ===')
train_data, val_data, norm_stats = load_data(
    '/mnt/shared-workspace/shared/preprocessing/batch_corrected_matrix.npy',
    '/workspace/conditioning_matrix.npy',
    '/mnt/shared-workspace/shared/preprocessing/sample_metadata_final.csv',
    val_frac=0.15, seed=42
)

# Load best model
print('=== Loading best model ===')
model = CVAE(n_genes=20869, cond_dim=33, latent_dim=32, beta=0.5, dropout=0.1)
checkpoint = torch.load('/workspace/cvae_model/best_model.pt', weights_only=False)
model.load_state_dict(checkpoint['model_state_dict'])
model.eval()
print(f'Loaded epoch {checkpoint["epoch"]}, val_loss={checkpoint["val_loss"]:.1f}')

# Extract latent embeddings for ALL samples
print('\n=== Extracting latent embeddings ===')
all_expr = np.vstack([train_data['expr'], val_data['expr']])
all_cond = np.vstack([train_data['cond'], val_data['cond']])
all_meta = pd.concat([train_data['meta'], val_data['meta']], ignore_index=True)

with torch.no_grad():
    x = torch.from_numpy(all_expr).float()
    c = torch.from_numpy(all_cond).float()
    mu, logvar = model.encode(x)
    latent = mu.numpy()

print(f'Latent embeddings: {latent.shape}')
print(f'Latent range: [{latent.min():.2f}, {latent.max():.2f}]')
print(f'Latent mean: {latent.mean():.3f}, SD: {latent.std():.3f}')

# Save latent embeddings
np.save('/workspace/cvae_model/latent_embeddings.npy', latent)
all_meta.to_csv('/workspace/cvae_model/latent_metadata.csv', index=False)

# Run UMAP
print('\n=== Running UMAP ===')
reducer = umap.UMAP(n_neighbors=30, min_dist=0.3, n_components=2, random_state=42)
latent_umap = reducer.fit_transform(latent)
print(f'UMAP shape: {latent_umap.shape}')
np.save('/workspace/cvae_model/latent_umap.npy', latent_umap)

# --- Figure: UMAP colored by primary stimulus and tissue ---
print('\n=== Creating UMAP figures ===')

all_meta['primary_stimulus'] = all_meta['stimulus'].apply(
    lambda x: str(x).split(';')[0].strip() if pd.notna(x) else 'unknown'
)
all_meta['primary_tissue'] = all_meta['tissue'].apply(
    lambda x: str(x).split(';')[0].strip() if pd.notna(x) else 'unknown'
)

stim_colors = plt.cm.tab20(np.linspace(0, 1, 20))
tissue_colors = plt.cm.Set2(np.linspace(0, 1, 8))

fig, axes = plt.subplots(1, 2, figsize=(16, 7))

stimuli = sorted(all_meta['primary_stimulus'].unique())
stim_cmap = {s: stim_colors[i % 20] for i, s in enumerate(stimuli)}
for s in stimuli:
    mask = all_meta['primary_stimulus'] == s
    axes[0].scatter(latent_umap[mask, 0], latent_umap[mask, 1],
                    c=[stim_cmap[s]], s=5, alpha=0.6, label=s)
axes[0].set_title('Latent UMAP by Stimulus', fontsize=14)
axes[0].set_xlabel('UMAP 1', fontsize=12)
axes[0].set_ylabel('UMAP 2', fontsize=12)
axes[0].legend(fontsize=7, markerscale=2, loc='best', framealpha=0.7)

tissues = sorted(all_meta['primary_tissue'].unique())
tissue_cmap = {t: tissue_colors[i % 8] for i, t in enumerate(tissues)}
for t in tissues:
    mask = all_meta['primary_tissue'] == t
    axes[1].scatter(latent_umap[mask, 0], latent_umap[mask, 1],
                    c=[tissue_cmap[t]], s=5, alpha=0.6, label=t)
axes[1].set_title('Latent UMAP by Tissue', fontsize=14)
axes[1].set_xlabel('UMAP 1', fontsize=12)
axes[1].set_ylabel('UMAP 2', fontsize=12)
axes[1].legend(fontsize=8, markerscale=2, loc='best', framealpha=0.7)

plt.tight_layout()
plt.savefig('/mnt/results/figures/fig2_latent_umap_pilot.svg', format='svg', dpi=150)
plt.savefig('/mnt/results/figures/fig2_latent_umap_pilot.png', format='png', dpi=150)
plt.close()
print('Saved: fig2_latent_umap_pilot.svg/png')

# --- Figure: Training loss curves ---
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

with open('/workspace/cvae_model/training_history.json') as f:
    hist = json.load(f)

axes[0].plot(hist['epochs'], hist['train_loss'], label='Train', color='#0279EE')
axes[0].plot(hist['epochs'], hist['val_loss'], label='Val', color='#FF9400')
axes[0].set_title('Total Loss', fontsize=14)
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].legend()
axes[0].set_yscale('log')

axes[1].plot(hist['epochs'], hist['train_recon'], label='Train', color='#0279EE')
axes[1].plot(hist['epochs'], hist['val_recon'], label='Val', color='#FF9400')
axes[1].set_title('Reconstruction Loss', fontsize=14)
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('MSE')
axes[1].legend()

axes[2].plot(hist['epochs'], hist['train_kl'], label='Train', color='#0279EE')
axes[2].plot(hist['epochs'], hist['val_kl'], label='Val', color='#FF9400')
axes[2].set_title('KL Divergence', fontsize=14)
axes[2].set_xlabel('Epoch')
axes[2].set_ylabel('KL')
axes[2].legend()

plt.tight_layout()
plt.savefig('/mnt/results/figures/fig_training_curves_pilot.svg', format='svg', dpi=150)
plt.savefig('/mnt/results/figures/fig_training_curves_pilot.png', format='png', dpi=150)
plt.close()
print('Saved: fig_training_curves_pilot.svg/png')

# --- Summary statistics ---
print('\n=== Latent Space Evaluation ===')
from sklearn.metrics import silhouette_score

stim_labels = all_meta['primary_stimulus'].values
tissue_labels = all_meta['primary_tissue'].values

valid_stim = pd.Series(stim_labels).value_counts()
valid_stim = valid_stim[valid_stim >= 10].index
mask_stim = np.isin(stim_labels, valid_stim)
if mask_stim.sum() > 100 and len(valid_stim) > 1:
    sil_stim = silhouette_score(latent[mask_stim], stim_labels[mask_stim])
    print(f'Silhouette score (stimulus, {len(valid_stim)} groups): {sil_stim:.3f}')

valid_tissue = pd.Series(tissue_labels).value_counts()
valid_tissue = valid_tissue[valid_tissue >= 10].index
mask_tissue = np.isin(tissue_labels, valid_tissue)
if mask_tissue.sum() > 100 and len(valid_tissue) > 1:
    sil_tissue = silhouette_score(latent[mask_tissue], tissue_labels[mask_tissue])
    print(f'Silhouette score (tissue, {len(valid_tissue)} groups): {sil_tissue:.3f}')

is_spaceflight = (stim_labels == 'spaceflight').astype(int)
if is_spaceflight.sum() > 10 and (1-is_spaceflight).sum() > 10:
    sil_sf = silhouette_score(latent, is_spaceflight)
    print(f'Silhouette score (spaceflight vs ground): {sil_sf:.3f}')

latent_active = (np.abs(latent.mean(axis=0)) > 0.1) | (latent.std(axis=0) > 0.1)
print(f'Active latent dimensions: {latent_active.sum()}/{latent.shape[1]}')
print(f'Latent per-dim mean: {latent.mean(axis=0).round(2)}')
print(f'Latent per-dim std: {latent.std(axis=0).round(2)}')

print('\n=== Evaluation Complete ===')
```
